## 一、作業介紹

這次我用 Python 做了一個簡單的聚餐分帳工具，主要是想知道大家一起吃飯時，算上服務費和折扣後，每個人實際需要付多少錢。

程式會先整理輸入的資料，再計算整筆餐費，最後把金額平均分給每個人，也會幫忙判斷有沒有超出原本設定的預算。

## 二、程式使用方式

先輸入餐費、人數、服務費比例和折扣金額，接著執行程式，就能看到計算後的結果。

## 三、測試內容

為了確認程式有沒有算錯，我分別調整了聚餐人數和折扣金額，測試三種不同的情況。

- **測試一：** 4 個人吃飯，使用 100 元折價券。
- **測試二：** 改成 5 個人分攤，折扣維持 100 元。
- **測試三：** 改回 4 個人，折扣增加到 220 元。

透過這三次測試，可以比較人數和折扣不同時，每個人需要付的金額會有什麼變化。

```python

# HW08 聚餐費用試算器

event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300
```

```python

# 清理聚餐名稱
clean_event_name = event_name.strip().lower()

# 清理餐費文字
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()

# 轉換資料型別
meal_value = float(clean_meal_text)
people_count = int(people)

# 顯示結果
print("聚餐名稱：", clean_event_name)
print("餐費：", meal_value)
print("人數：", people_count)
```

```python

# 計算服務費
service_fee = meal_value * service_rate

# 計算折扣後總金額
total_amount = meal_value + service_fee - discount

# 計算每人應付金額
amount_per_person = total_amount / people_count

# 判斷是否超過預算
is_over_budget = amount_per_person > budget_per_person

# 顯示結果
print("服務費：", service_fee)
print("折扣：", discount)
print("應付總額：", total_amount)
print("每人費用：", amount_per_person)
print("是否超過預算：", is_over_budget)
```

```python

# 第二組測試：5人，折扣100元

people = "5"
discount = 100

# 轉換人數
people_count = int(people)

# 重新計算
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count
is_over_budget = amount_per_person > budget_per_person

# 顯示結果
print("第二組測試")
print("人數：", people_count)
print("折扣：", discount)
print("每人費用：", amount_per_person)
print("是否超過預算：", is_over_budget)
```

```python

# 第三組測試：4人，折扣220元

people = "4"
discount = 220

# 轉換人數
people_count = int(people)

# 重新計算
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count
is_over_budget = amount_per_person > budget_per_person

# 顯示結果
print("第三組測試")
print("人數：", people_count)
print("折扣：", discount)
print("每人費用：", amount_per_person)
print("是否超過預算：", is_over_budget)
```
