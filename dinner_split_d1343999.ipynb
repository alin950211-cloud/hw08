```python

# HW08 聚餐費用試算器

event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300
```

```python

# 清理聚餐名稱
clean_event_name = event_name.strip().lower()

# 清理餐費文字
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()

# 轉換資料型別
meal_value = float(clean_meal_text)
people_count = int(people)

# 顯示結果
print("聚餐名稱：", clean_event_name)
print("餐費：", meal_value)
print("人數：", people_count)
```

```python

# 計算服務費
service_fee = meal_value * service_rate

# 計算折扣後總金額
total_amount = meal_value + service_fee - discount

# 計算每人應付金額
amount_per_person = total_amount / people_count

# 判斷是否超過預算
is_over_budget = amount_per_person > budget_per_person

# 顯示結果
print("服務費：", service_fee)
print("折扣：", discount)
print("應付總額：", total_amount)
print("每人費用：", amount_per_person)
print("是否超過預算：", is_over_budget)
```

```python

# 第二組測試：5人，折扣100元

people = "5"
discount = 100

# 轉換人數
people_count = int(people)

# 重新計算
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count
is_over_budget = amount_per_person > budget_per_person

# 顯示結果
print("第二組測試")
print("人數：", people_count)
print("折扣：", discount)
print("每人費用：", amount_per_person)
print("是否超過預算：", is_over_budget)
```
